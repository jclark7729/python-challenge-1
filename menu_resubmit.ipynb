```python
# Menu dictionary
menu = {
    "Snacks": {
        "Cookie": .99,
        "Banana": .69,
        "Apple": .49,
        "Granola bar": 1.99
    },
    "Meals": {
        "Burrito": 4.49,
        "Teriyaki Chicken": 9.99,
        "Sushi": 7.49,
        "Pad Thai": 6.99,
        "Pizza": {
            "Cheese": 8.99,
            "Pepperoni": 10.99,
            "Vegetarian": 9.99
        },
        "Burger": {
            "Chicken": 7.49,
            "Beef": 8.49
        }
    },
    "Drinks": {
        "Soda": {
            "Small": 1.99,
            "Medium": 2.49,
            "Large": 2.99
        },
        "Tea": {
            "Green": 2.49,
            "Thai iced": 3.99,
            "Irish breakfast": 2.49
        },
        "Coffee": {
            "Espresso": 2.99,
            "Flat white": 2.99,
            "Iced": 3.49
        }
    },
    "Dessert": {
        "Chocolate lava cake": 10.99,
        "Cheesecake": {
            "New York": 4.99,
            "Strawberry": 6.49
        },
        "Australian Pavlova": 9.99,
        "Rice pudding": 4.99,
        "Fried banana": 4.49
    }
}

# 1. Set up order list
order_list = []

# Launch the store and present a greeting to the customer
customer_greeting = "Welcome to the Best Deli!"
print(customer_greeting)

# Main loop for ordering
loop_order = True
while loop_order:
    print("\nPlease select a menu category.")

    # Print the menu categories
    menu_item_number = 1
    menu_dict = {}
    for menu_category in menu:
        print(f"{menu_item_number}. {menu_category}")
        menu_dict[menu_item_number] = menu_category
        menu_item_number += 1

    customer_input_str = input("Enter the number of the category: ")

    if customer_input_str.isdigit():
        customer_input = int(customer_input_str)
        if customer_input in menu_dict:
            menu_category_name = menu_dict[customer_input]
            print(f"You selected {menu_category_name}.")

            # Loop for item selection within the category
            select_item = True
            while select_item:
                print(f"\nItems in {menu_category_name}:")
                item_number = 1
                for item_name, item_price in menu[menu_category_name].items():
                    if isinstance(item_price, dict):
                        # Handle subcategories (like Pizza or Burger)
                        for sub_item, sub_price in item_price.items():
                            print(f"{item_number}. {item_name} - {sub_item}: ${sub_price:.2f}")
                            item_number += 1
                    else:
                        print(f"{item_number}. {item_name}: ${item_price:.2f}")
                        item_number += 1

                item_selection = input("Please select an item number (or 'back' to return to categories): ")

                if item_selection.lower() == "back":
                    select_item = False  # Exit item selection loop, go back to categories
                    break

                if item_selection.isdigit():
                    item_selection = int(item_selection)
                    item_number = 1
                    valid_selection = False
                    for item_name, item_price in menu[menu_category_name].items():
                        if isinstance(item_price, dict):
                            for sub_item, sub_price in item_price.items():
                                if item_number == item_selection:
                                    menu_item_name = f"{item_name} - {sub_item}"
                                    menu_item_price = sub_price
                                    valid_selection = True
                                    break
                                item_number += 1
                        else:
                            if item_number == item_selection:
                                menu_item_name = item_name
                                menu_item_price = item_price
                                valid_selection = True
                            item_number += 1

                    if valid_selection:
                        quantity_str = input(f"How many {menu_item_name}s would you like? ")

                        if quantity_str.isdigit():
                            quantity = int(quantity_str)
                        else:
                            quantity = 1

                        order_list.append({
                            "name": menu_item_name,
                            "price": menu_item_price,
                            "quantity": quantity
                        })

                        print(f"Added {quantity} x {menu_item_name} to your order.")
                        select_item = False  # Exit item selection loop after successful item addition
                    else:
                        print("Invalid item number. Please try again.")
                else:
                    print("Invalid input. Please enter a number for item selection or 'back'.")

        else:
            print("Invalid category. Please enter a valid number.")
    else:
        print("Invalid input. Please enter a number.")

    another_item = input("Do you want to order another item? (yes/no): ")
    if another_item.lower() != 'yes':
        loop_order = False

# Calculate the total
total = sum(item["price"] * item["quantity"] for item in order_list)

# Print out the customer's order
print("\nThank you for your order. Here is your order:")
for order in order_list:
    menu_item_name = order["name"]
    price = order["price"]
    quantity = order["quantity"]

    # Calculate the number of spaces for formatted printing
    spaces = " " * (20 - len(menu_item_name))
    spaces_price = " " * (20 - len(str(price)))
    spaces_quantity = " " * (20 - len(str(quantity)))

    # Print the item name, price, and quantity
    print(f"{menu_item_name}{spaces}{price:.2f}{spaces_price}{quantity}{spaces_quantity}")
    print(f"{'-' * 20}")

# Print the total cost
print(f"Your total cost is: ${total:.2f}")
print("Thank you for your order. Have a great day!")
```
